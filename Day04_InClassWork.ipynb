```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

#read the csv file
db = pd.read_csv("/content/house_price_regression_dataset.csv")
db.info()

#get the input and output
x = db[["Square_Footage", "Num_Bedrooms", "Num_Bathrooms", "Lot_Size", ]]
y = db["House_Price"]

#train the model
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.20, random_state = 42)

#create a model
model = LinearRegression()

#training
train_model = model.fit(x_train, y_train)

#predict
predict = model.predict(x_test)

prediction = model.predict([[1500, 2, 2, 15]])
print(prediction)

#gettting the
from sklearn.metrics import r2_score, mean_squared_error
validation = r2_score(y_test, predict)
print(validation)

mae = mean_squared_error(y_test, predict)
print("MAE: ", mae)

mse = mean_squared_error(y_test, predict)
print("MSE: ", mse)

```
