```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

df = pd.read_csv("/content/study_hours_marks_100.csv")

x = df[["Study_Hours"]]
y = df["Marks"]

#splitting the dataset into 80:20  ration 80-train and 10-testing from (90:10, 70:30)
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = 0.20, random_state = 42)

model = LinearRegression()
model.fit(x_train, y_train)

#to test the model (testing)
predictions = model.predict(x_test)
print(predictions)

#to see how perfect are my predictions (evaluation)
"""
1.0 - perfect prediciton
0.9 model explains over 90% of the variations
0.8 model explains over 80% of the variations
0.5 model explains over 50% of the variations
0.0 No better than prediction the average
< 0 Worse than simply using the average value
"""
score = r2_score(y_test, predictions)
print(score)
```
